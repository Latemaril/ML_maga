### Основные признаки
**age** – возраст пациента в годах.

**sex** – пол: 1 – мужчина, 0 – женщина.

**cp** – тип боли в груди (chest pain type): 0 – типичная стенокардия, 1 – атипичная стенокардия, 2 – неангинозная боль, 3 – бессимптомно.  ​

### Давление и холестерин
**trestbps** – артериальное давление в состоянии покоя (resting blood pressure) в мм рт. ст

**chol** – уровень холестерина в сыворотке крови, мг/дл.

**fbs** – уровень сахара натощак > 120 мг/дл: 1 – да, 0 – нет.

### ЭКГ и пульс
**restecg** – результаты ЭКГ в покое: 0 – признаки гипертрофии левого желудочка, 1 – нормальная ЭКГ, 2 – изменения ST–T.

**thalach** – максимальная достигнутая частота пульса.

### Нагрузка и сегмент ST
**exang** – приступ стенокардии, вызванный нагрузкой: 1 – да, 0 – нет.

**oldpeak** – депрессия сегмента ST при нагрузке относительно покоя (в единицах отклонения).

**slope** – форма (наклон) пикового сегмента ST: 0 – нисходящий, 1 – горизонтальный, 2 – восходящий.

### Сосуды, талассемия и цель
**ca** – число крупных сосудов (0–3), окрашенных при флюороскопии.

**thal** – состояние талассемии: обычно 1 – фиксированный дефект, 2 – нормальное, 3 – обратимый дефект.

**target** – наличие болезни сердца: 1 – есть заболевание, 0 – нет.

## Классификация переменных по типам

### Номинальные (категориальные, без порядка)
**sex** – пол: 0/1, просто разные категории, порядка нет.

**fbs** – сахар натощак > 120 мг/дл: 0/1, просто “нет/да”.

**exang** – стенокардия от нагрузки: 0/1, “нет/да”.

**restecg** – тип ЭКГ в покое: три разных состояния, логического порядка нет.

**ca** – число окрашенных сосудов: 0–3, это счётчик категорий, а не непрерывная шкала.

**thal** – тип талассемии (нормальная/фиксированный дефект/обратимый дефект).

**target** – наличие болезни: 0 – нет, 1 – есть. ​

### Порядковые (есть ранг, но мат. операции условны)
**cp** – тип боли в груди: 0–3, это ранги видов боли (типичная/атипичная и т.д.), можно говорить “сильнее/слабее”, но усреднять особого смысла нет.

**slope** – наклон ST сегмента: 0 – нисходящий, 1 – плоский, 2 – восходящий; естественный порядок по “улучшению” формы ST.

### Числовые непрерывные / количественные
**age** – возраст в годах, есть порядок, сумма/среднее имеют смысл.

**trestbps** – давление в покое, числовая шкала, можно применять мат. операции.

**chol** – холестерин, непрерывная величина.

**thalach** – максимальный пульс, непрерывная количественная.

**oldpeak** – депрессия ST, обычно вещественное число, тоже количественная

```python
# список количественных (числовых) переменных
print("ЧИСЛОВЫЕ ПЕРЕМЕННЫЕ:")
num_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']

# если в датафрейме они не все float/int – приведём:
df[num_cols] = df[num_cols].astype('float')

# описательная статистика только по числовым
num_stats = df[num_cols].describe()  # count, mean, std, min, 25%, 50%, 75%, max
print(num_stats)
```

```python
# номинальные + порядковые как категориальные
print("\nКАТЕГОРИАЛЬНЫЕ ПЕРЕМЕННЫЕ:")
cat_cols = ['sex', 'fbs', 'exang', 'restecg', 'ca', 'thal', 'target', 'cp', 'slope']

# приведём к категориальному типу
for c in cat_cols:
    df[c] = df[c].astype('category')

# описательная статистика по категориальным
cat_stats = df[cat_cols].describe()  # count, unique, top, freq
print(cat_stats)
```
